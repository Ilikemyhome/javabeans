Using a Python dictionary or a NumPy array format, construct a Pandas DataFrame named coffee_sales.

```python
import pandas as pd

data = {
    "Americano": [5628, 6239],
    "Latte": [12426, 25384],
    "Espresso": [852, 756],
    "Cappuccino": [8522, 7561],
}

index_labels = ["2021 Sales", "2022 Sales"]

coffee_sales = pd.DataFrame(data, index = index_labels)

coffee_sales.index.name = "Years (Index)"

coffee_sales
```

Calculate Total Product Volume: 

```python
# Find the cumulative sales for each distinct beverage type across the entire two-year timespan. Append or display this total safely.
total_sales = coffee_sales.sum()

total_sales
```

Year-over-Year (YoY) Growth: Calculate the percentage growth rate for each beverage type from 2021 to 2022.

```python
yoy_growth = ((coffee_sales.loc["2022 Sales"] - coffee_sales.loc["2021 Sales"]) 
              / coffee_sales.loc["2021 Sales"]) * 100
yoy_growth
```

Identify the absolute best-selling and least-selling beverage for both 2021 and 2022.

```python
best_2021 = coffee_sales.loc["2021 Sales"].idxmax()
least_2021 = coffee_sales.loc["2021 Sales"].idxmin()

best_2021, least_2021



```

```python
best_2022 = coffee_sales.loc["2022 Sales"].idxmax()
least_2022 = coffee_sales.loc["2022 Sales"].idxmin()

best_2022, least_2022
```
